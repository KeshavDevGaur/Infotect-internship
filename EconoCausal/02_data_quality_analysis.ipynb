```python
from pathlib import Path
import pandas as pd
import numpy as np

DATA_DIR = Path("../data/raw")

data = {
    file.stem: pd.read_csv(file)
    for file in DATA_DIR.glob("*.csv")
}

for name, df in data.items():
    print(f"{name}: {df.shape}")
```

```python
missing_summary = []

for name, df in data.items():
    for column in df.columns:
        missing_count = df[column].isna().sum()

        if missing_count > 0:
            missing_summary.append({
                "dataset": name,
                "column": column,
                "missing_count": missing_count,
                "missing_percentage": round(
                    missing_count / len(df) * 100, 2
                )
            })

missing_df = pd.DataFrame(missing_summary)

if not missing_df.empty:
    display(
        missing_df.sort_values(
            "missing_percentage",
            ascending=False
        ).reset_index(drop=True)
    )
else:
    print("No missing values found.")
```

```python
duplicate_summary = []

for name, df in data.items():
    duplicate_summary.append({
        "dataset": name,
        "total_rows": len(df),
        "duplicate_rows": int(df.duplicated().sum()),
        "duplicate_percentage": round(
            df.duplicated().mean() * 100, 2
        )
    })

duplicate_df = pd.DataFrame(duplicate_summary)

display(
    duplicate_df.sort_values(
        "duplicate_rows",
        ascending=False
    ).reset_index(drop=True)
)
```

```python
print("Redemption status counts:")
display(data["train"]["redemption_status"].value_counts(dropna=False))

print("\nRedemption status percentages:")
display(
    data["train"]["redemption_status"]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
)
```

```python
print("Redemption status counts:")
display(data["train"]["redemption_status"].value_counts(dropna=False))

print("\nRedemption status percentages:")
display(
    data["train"]["redemption_status"]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
)
```

```python
print("Campaign ID coverage:")
print(
    data["train"]["campaign_id"].isin(
        data["campaign_data"]["campaign_id"]
    ).value_counts()
)

print("\nCustomer ID coverage:")
print(
    data["train"]["customer_id"].isin(
        data["customer_demographics"]["customer_id"]
    ).value_counts()
)

print("\nCoupon ID coverage:")
print(
    data["train"]["coupon_id"].isin(
        data["coupon_item_mapping"]["coupon_id"]
    ).value_counts()
)
```

```python
customer_match = data["train"]["customer_id"].isin(
    data["customer_demographics"]["customer_id"]
)

print("Training records:", len(data["train"]))
print("Records with demographic matches:", customer_match.sum())
print("Records without demographic matches:", (~customer_match).sum())
print("Percentage with matches:", round(customer_match.mean() * 100, 2), "%")
print("Percentage without matches:", round((~customer_match).mean() * 100, 2), "%")
```
