```python
import pandas as pd
import numpy as np

print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
```

```python
from pathlib import Path

DATA_DIR = Path("../data/raw")

print("Data folder exists:", DATA_DIR.exists())
print("\nCSV files found:")

for file in sorted(DATA_DIR.glob("*.csv")):
    print(file.name)
```

```python
files = {
    "train": "train.csv",
    "test": "test_QyjYwdj.csv",
    "campaigns": "campaign_data.csv",
    "customers": "customer_demographics.csv",
    "transactions": "customer_transaction_data.csv",
    "coupon_mapping": "coupon_item_mapping.csv",
    "items": "item_data.csv",
    "submission": "sample_submission_Byiv0dS.csv"
}

data = {}

for name, filename in files.items():
    path = DATA_DIR / filename

    if path.exists():
        data[name] = pd.read_csv(path)
        print(f"{name}: {data[name].shape}")
    else:
        print(f"Missing file: {filename}")
```

```python
data["train"]
```

```python
summary = []

for name, df in data.items():
    summary.append({
        "dataset": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "missing_cells": int(df.isna().sum().sum()),
        "duplicate_rows": int(df.duplicated().sum())
    })

summary_df = pd.DataFrame(summary)

display(summary_df)
```

```python
summary_df.to_csv("../reports/day01_dataset_summary.csv", index=False)

print("Dataset summary saved successfully!")
```
