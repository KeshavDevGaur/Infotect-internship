## Step 2: Original Dataset Documentation Research

### Dataset Source

AmExpert 2019 Machine Learning Hackathon, organized by American Express and Analytics Vidhya.

### Problem Statement

The original challenge focused on predicting coupon redemption for customer-coupon combinations using historical campaign performance, customer demographics, product information, and transaction data.

### Documented Findings

1. Customers receive coupons under different marketing campaigns.
2. Coupons are applicable to specific products or product ranges.
3. Coupon redemption takes place within the campaign period.
4. Coupon redemption is reflected in transaction data through the `coupon_discount` variable.
5. The training data contains coupon-customer combinations from 18 campaigns.
6. Campaign metadata contains 28 campaigns with anonymized campaign types X and Y.

### External Research

A peer-reviewed 2023 PLOS ONE study by Henrika Langen and Martin Huber used the AmExpert dataset to investigate causal effects of coupon provision on retailer sales.

The study applied causal machine learning and emphasized the importance of accounting for factors influencing both coupon provision and purchasing behavior.

### Important Research Limitations

* The original documentation does not explain the actual meanings of campaign types X and Y.
* The precise customer-level campaign assignment mechanism is not documented in the available problem statement.
* The training records represent coupon-customer impressions, not an explicit sample of customers receiving no coupons.
* Coupon redemption is not equivalent to the causal effect of coupon provision.
* Additional investigation is required to determine whether the dataset supports the intended causal estimand.

### Initial Conclusion

The original documentation establishes that customers receive coupons under marketing campaigns and may redeem them. However, the exact assignment mechanism and the interpretation of campaign types X and Y remain unresolved.

Further research is required before defining a defensible treatment variable for Double Machine Learning.


## Step 3: Research Paper Methodology Review

### Research Paper

How causal machine learning can leverage marketing strategies: Assessing and improving the performance of a coupon campaign (2023).

**Authors:** Henrika Langen and Martin Huber.

### Methodology Used

The researchers applied causal forest methodology to estimate the effects of coupon provision on retailer sales.

### Main Research Components

1. Treatment: Coupon provision.
2. Outcome: Retailer sales.
3. Covariates: Customer characteristics and pre-campaign purchasing information.
4. ATE: Average effect of coupon provision.
5. CATE: Variation in coupon effects across customer subgroups.
6. Policy Learning: Identifying customer groups that could benefit most from coupon provision.

### Important Assumption

The study relies on an unconfoundedness or selection-on-observables assumption: relevant factors jointly influencing coupon reception and purchasing behavior must be observed and appropriately controlled for.

### Relevance to EconoCausal

The paper demonstrates how causal machine learning can be applied to coupon campaign evaluation.

However, its methodology cannot automatically be transferred to our project. We must independently verify the available treatment information, outcome measurements, covariates, and identification assumptions.

### Predictive ML vs Causal ML

Predictive ML estimates likely outcomes based on observed patterns.

Causal ML estimates intervention effects using counterfactual reasoning and appropriate identification assumptions.

### Conclusion

The research provides a useful methodological reference for EconoCausal. Causal Forest and Double Machine Learning are potential approaches, but the appropriate method depends on the research question, data structure, and identification feasibility.


## Step 4: Treatment and Outcome Definition Review

### Reference Research

Langen and Huber (2023), "How causal machine learning can leverage marketing strategies."

### Treatment Definition

The reference study defines treatment as customer exposure to a coupon campaign.

* D = 1: Customer receives a coupon of the specified type.
* D = 0: Customer does not receive that coupon type.

Coupon redemption is not the treatment definition.

### Outcome Definition

The reference study uses average per-day customer expenditure during the campaign period as the outcome variable.

### Covariates

The study considers:

* Customer socioeconomic characteristics.
* Previous average daily spending by product type.
* Previous coupon reception.
* Previous coupon redemption.
* Other coupon types received during the current campaign period.

### Comparison with EconoCausal

Our current training dataset contains customer-coupon records and redemption information. However, the intended treatment variable and expenditure outcome must be constructed and verified independently.

### Important Research Gap

The current project has not yet established whether the available records provide a valid customer-level comparison between coupon recipients and non-recipients.

### Conclusion

The reference paper provides a useful framework for defining treatment, outcome, and covariates. Its methodology should not be copied directly without verifying that our dataset supports the same research design.


```python
# Step 5: Transaction data investigation

import pandas as pd
import numpy as np
transaction_data = pd.read_csv(
    "../data/raw/customer_transaction_data.csv"
)

# Inspect dataset dimensions
print("Transaction dataset shape:", transaction_data.shape)

# Inspect columns
print("\nAvailable columns:")
print(transaction_data.columns.tolist())

# Display sample records
print("\nSample transaction records:")
display(transaction_data.head())

# Check data types
print("\nData types:")
print(transaction_data.dtypes)

# Check missing values
print("\nMissing values:")
display(
    transaction_data.isnull().sum().to_frame(
        name="missing_count"
    )
)

# Check customer and item identifiers
for col in ["customer_id", "item_id"]:
    if col in transaction_data.columns:
        print(
            f"\nUnique {col}:",
            transaction_data[col].nunique()
        )
```

```python
print("Columns:")
print(transaction_data.columns.tolist())

print("\nFirst 5 records:")
display(transaction_data.head())

print("\nData types:")
print(transaction_data.dtypes)
```

```python
# Step 6: Transaction variable analysis

# Convert date into datetime
transaction_data['date'] = pd.to_datetime(
    transaction_data['date']
)

# Dataset date range
print("Minimum date:", transaction_data['date'].min())
print("Maximum date:", transaction_data['date'].max())

# Numerical summary
print("\nNumerical Summary:")
display(transaction_data.describe())

# Discount and quantity analysis
print("\nQuantity distribution:")
display(transaction_data['quantity'].value_counts().head(10))

print("\nDiscount summary:")
display(
    transaction_data[
        ['selling_price', 'other_discount', 'coupon_discount']
    ].describe()
)

# Check missing values
print("\nMissing values:")
display(transaction_data.isnull().sum())
```

```python
# Inspect price and discount relationships

transaction_data['gross_value'] = (
    transaction_data['quantity'] *
    transaction_data['selling_price']
)

print("Total transaction records:", len(transaction_data))

print("\nTotal gross value:")
print(transaction_data['gross_value'].sum())

print("\nTotal other discount:")
print(transaction_data['other_discount'].sum())

print("\nTotal coupon discount:")
print(transaction_data['coupon_discount'].sum())

print("\nSample calculated values:")
display(
    transaction_data[
        ['quantity', 'selling_price',
         'gross_value', 'other_discount',
         'coupon_discount']
    ].head(10)
)
```

```python
# Step 8: Investigate transaction value distribution

print("Quantity statistics:")
display(transaction_data['quantity'].describe(
    percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
))

print("\nSelling price statistics:")
display(transaction_data['selling_price'].describe())

print("\nLargest quantities:")
display(
    transaction_data.nlargest(10, 'quantity')[
        ['date', 'customer_id', 'item_id',
         'quantity', 'selling_price',
         'other_discount', 'coupon_discount']
    ]
)

print("\nLargest gross values:")
display(
    transaction_data.nlargest(10, 'gross_value')[
        ['date', 'customer_id', 'item_id',
         'quantity', 'selling_price',
         'gross_value', 'other_discount',
         'coupon_discount']
    ]
)
```

```python
# Step 9: Outlier investigation

print("Quantity above 100:")
print((transaction_data['quantity'] > 100).sum())

print("\nQuantity above 1000:")
print((transaction_data['quantity'] > 1000).sum())

print("\nSelling price above 1000:")
print((transaction_data['selling_price'] > 1000).sum())

print("\nQuantity percentiles:")
display(
    transaction_data['quantity'].quantile(
        [0.50, 0.75, 0.90, 0.95, 0.99, 0.995, 0.999]
    )
)

print("\nLargest transaction records:")
display(
    transaction_data.nlargest(15, 'quantity')
)
```

```python
# Step 10: Extreme quantity investigation

# Filter extreme quantity records
extreme_data = transaction_data[
    transaction_data['quantity'] > 1000
].copy()

print("Extreme records:", len(extreme_data))

print("\nUnique customers:")
print(extreme_data['customer_id'].nunique())

print("\nUnique items:")
print(extreme_data['item_id'].nunique())

print("\nTop 10 items with extreme quantities:")

display(
    extreme_data.groupby('item_id')
    .agg(
        records=('quantity', 'count'),
        total_quantity=('quantity', 'sum'),
        median_quantity=('quantity', 'median'),
        max_quantity=('quantity', 'max')
    )
    .sort_values('records', ascending=False)
    .head(10)
)

print("\nTop 10 customers with extreme quantities:")

display(
    extreme_data.groupby('customer_id')
    .agg(
        records=('quantity', 'count'),
        total_quantity=('quantity', 'sum'),
        max_quantity=('quantity', 'max')
    )
    .sort_values('records', ascending=False)
    .head(10)
)
```

```python
# Step 11: Investigate item metadata

item_data = pd.read_csv("../data/raw/item_data.csv")

print("Item dataset shape:", item_data.shape)

print("\nItem columns:")
print(item_data.columns.tolist())

# Inspect extreme items
extreme_item_ids = extreme_data['item_id'].unique()

extreme_items = item_data[
    item_data['item_id'].isin(extreme_item_ids)
]

print("\nExtreme item metadata:")
display(extreme_items.head(30))

print("\nNumber of matching items:", len(extreme_items))
```

```python
# Step 12: Category-wise transaction analysis

# Merge transaction data with item metadata
transaction_items = transaction_data.merge(
    item_data,
    on="item_id",
    how="left"
)

# Category-wise summary
category_summary = (
    transaction_items.groupby("category")
    .agg(
        transaction_count=("quantity", "count"),
        unique_customers=("customer_id", "nunique"),
        unique_items=("item_id", "nunique"),
        total_quantity=("quantity", "sum"),
        median_quantity=("quantity", "median"),
        average_quantity=("quantity", "mean")
    )
    .sort_values("transaction_count", ascending=False)
)

display(category_summary)

print("\nExtreme quantity records by category:")

display(
    transaction_items[
        transaction_items["quantity"] > 1000
    ]["category"].value_counts(dropna=False)
)
```

```python
# Step 13: Category-wise monetary analysis

monetary_summary = (
    transaction_items.groupby("category")
    .agg(
        transaction_count=("selling_price", "count"),
        median_selling_price=("selling_price", "median"),
        mean_selling_price=("selling_price", "mean"),
        median_other_discount=("other_discount", "median"),
        median_coupon_discount=("coupon_discount", "median"),
        total_other_discount=("other_discount", "sum"),
        total_coupon_discount=("coupon_discount", "sum")
    )
    .sort_values("transaction_count", ascending=False)
)

display(monetary_summary)

print("\nFuel transaction examples:")

display(
    transaction_items[
        transaction_items["category"] == "Fuel"
    ][
        ["date", "customer_id", "item_id",
         "quantity", "selling_price",
         "other_discount", "coupon_discount"]
    ].head(10)
)

print("\nGrocery transaction examples:")

display(
    transaction_items[
        transaction_items["category"] == "Grocery"
    ][
        ["date", "customer_id", "item_id",
         "quantity", "selling_price",
         "other_discount", "coupon_discount"]
    ].head(10)
)
```

```python
# Step 14: Monetary relationship investigation

transaction_data['gross_value'] = (
    transaction_data['quantity'] *
    transaction_data['selling_price']
)

print("Discount sign analysis:")

for col in ['other_discount', 'coupon_discount']:
    print(f"\n{col}")
    print("Negative:", (transaction_data[col] < 0).sum())
    print("Zero:", (transaction_data[col] == 0).sum())
    print("Positive:", (transaction_data[col] > 0).sum())

print("\nSelling price per quantity summary:")

display(
    (
        transaction_data['selling_price'] /
        transaction_data['quantity'].replace(0, np.nan)
    ).describe()
)

print("\nTransaction examples with coupon discounts:")

display(
    transaction_data[
        transaction_data['coupon_discount'] < 0
    ].head(15)
)
```

```python
# Step 15: Selling price and quantity relationship

print("Correlation between quantity and selling price:")

display(
    transaction_data[
        ['quantity', 'selling_price']
    ].corr()
)

print("\nTransactions with quantity = 1:")

display(
    transaction_data[
        transaction_data['quantity'] == 1
    ].head(10)
)

print("\nTransactions with quantity > 1 and < 10:")

display(
    transaction_data[
        (transaction_data['quantity'] > 1) &
        (transaction_data['quantity'] < 10)
    ].head(10)
)

print("\nTransactions with quantity > 1000:")

display(
    transaction_data[
        transaction_data['quantity'] > 1000
    ].head(10)
)
```

```python
# Step 16: Detailed investigation of item 49009

item_49009 = transaction_items[
    transaction_items['item_id'] == 49009
].copy()

print("Item 49009 records:", len(item_49009))

print("\nQuantity statistics:")
display(item_49009['quantity'].describe())

print("\nSelling price statistics:")
display(item_49009['selling_price'].describe())

print("\nQuantity and selling price correlation:")
print(
    item_49009[['quantity', 'selling_price']].corr()
)

print("\nSample records:")
display(
    item_49009[
        ['date', 'customer_id', 'quantity',
         'selling_price', 'other_discount',
         'coupon_discount']
    ].head(20)
)

print("\nItem metadata:")
display(
    item_data[item_data['item_id'] == 49009]
)
```

```python
# Step 17: Price-to-quantity ratio

item_49009['price_per_quantity'] = (
    item_49009['selling_price'] /
    item_49009['quantity']
)

print("Price per quantity statistics:")
display(item_49009['price_per_quantity'].describe())

print("\nSample price-to-quantity ratios:")
display(
    item_49009[
        ['quantity', 'selling_price', 'price_per_quantity']
    ].head(20)
)

print("\nCorrelation with quantity:")
print(
    item_49009[
        ['quantity', 'selling_price', 'price_per_quantity']
    ].corr()
)
```

```python
# Compare all Fuel items

fuel_items = transaction_data.merge(
    item_data[['item_id', 'category']],
    on='item_id',
    how='left'
)

fuel_items = fuel_items[
    fuel_items['category'] == 'Fuel'
].copy()

fuel_summary = fuel_items.groupby('item_id').agg(
    records=('quantity', 'count'),
    avg_quantity=('quantity', 'mean'),
    median_quantity=('quantity', 'median'),
    avg_selling_price=('selling_price', 'mean')
)

fuel_summary['price_per_quantity'] = (
    fuel_summary['avg_selling_price'] /
    fuel_summary['avg_quantity']
)

display(fuel_summary.sort_values(
    'records', ascending=False
))
```

```python
# Step 19: Filter frequently observed Fuel items

reliable_fuel = fuel_summary[
    fuel_summary['records'] >= 10
].copy()

print("Reliable Fuel items:")
display(reliable_fuel)

print("\nNumber of reliable Fuel items:")
print(len(reliable_fuel))

print("\nPrice-to-quantity ratio summary:")
print(
    reliable_fuel['price_per_quantity'].describe()
)
```
